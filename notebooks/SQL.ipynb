```python
import os
import psycopg2
from pgvector.psycopg2 import register_vector

DB_CONFIG = {
    "dbname": "FACEPROJECT",
    "user": "postgres",
    "password": os.environ.get("DB_PASSWORD", ""),  # Put your actual password here
    "host": "localhost",
    "port": 5432
}

try:
    conn = psycopg2.connect(**DB_CONFIG)
    register_vector(conn)
    cur = conn.cursor()
    cur.execute("SELECT COUNT(*) FROM missing_persons;")
    total = cur.fetchone()[0]
    print(f"[+] Connected to FACEPROJECT successfully! Records count: {total}")
    cur.close()
    conn.close()
except Exception as e:
    print(f"[!] Connection failed: {e}")
```
